```python
file_path = 'C:\\Users\\dip.das\\Desktop\\My Python Stuff\\Section 14\\Example Top Level'
```

```python
for folders, sub_folders, files in os.walk(file_path):
    print(f"Currently looking at {folders}")
    print('\n')
    print(f'The subfolders are:')
    for sub_folds in sub_folders:
        print(f"\tSubfolder: {sub_folds}")
    print('\n')
    print('The files are:')
    for file in files:
        print(f"\tFile: {file}")
    print('\n')
```
