```python
text = "My phone number is 555-408-1234"
```

```python
phone = re.search(r'\d{3}-\d{3}-\d{4}',text)
```

```python
phone
```

```python
phone.group()
```

```python
phone_pattern = re.compile(r'(\d{3})-(\d{3})-(\d{4})')
```

```python
results = re.search(phone_pattern,text)
```

```python
results
```

```python
results.group(3)
```
