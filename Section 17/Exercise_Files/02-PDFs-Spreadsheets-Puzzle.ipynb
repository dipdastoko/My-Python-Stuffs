## Task One: Grab the Google Drive Link from .csv File

```python
import csv
import requests
import pypdf
import gdown
import re
```

```python
data = open('find_the_link.csv')
```

```python
csv_data = csv.reader(data)
```

```python
data_lines = list(csv_data)
```

```python
len(data_lines)
```

```python
url=''
for index,item in enumerate(data_lines):
    url+=item[index]
```

## Task Two: Download the PDF from the Google Drive link and find the phone number that is in the document. 

```python
# You should get this phone number
# 505 503 4455
```

```python
gdown.download(url,"downloaded_pdf.pdf",quiet=False)
```

```python
f = open('downloaded_pdf.pdf','rb')
```

```python
pdf_reader = pypdf.PdfReader(f)
```

```python
pdf_reader.get_num_pages()
```

```python
pdf_text = ''
for pageNo in range(pdf_reader.get_num_pages()):
    pagetext = pdf_reader.get_page(pageNo).extract_text()
    pdf_text += pagetext
```

```python
phoneNumbers = re.findall('\d{3}\.\d{3}\.\d{4}',pdf_text)
```
